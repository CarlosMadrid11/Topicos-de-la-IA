```python
import os

BASE = '/content/drive/MyDrive/reconocimiento_facial_tec_culiacan'

carpetas = [
    f'{BASE}/dataset/raw',
    f'{BASE}/dataset/augmented',
    f'{BASE}/dataset/processed/train',
    f'{BASE}/dataset/processed/val',
    f'{BASE}/dataset/processed/test',
    f'{BASE}/modelo',
    f'{BASE}/metricas',
    f'{BASE}/app_data',
]

for carpeta in carpetas:
    os.makedirs(carpeta, exist_ok=True)

print('Estructura de carpetas creada:')
for carpeta in carpetas:
    print(f'    {carpeta.replace(BASE, "reconocimiento_facial_tec_culiacan")}')
```

```python
import subprocess

result = subprocess.run(['pip', 'freeze'], capture_output=True, text=True)
requirements = result.stdout

ruta = f'{BASE}/requirements.txt'
with open(ruta, 'w') as f:
    f.write(requirements)

print(f' requirements.txt guardado en Drive')
print(f'   Ruta: {ruta}')
print(f'   Total de dependencias registradas: {len(requirements.splitlines())}')
```
