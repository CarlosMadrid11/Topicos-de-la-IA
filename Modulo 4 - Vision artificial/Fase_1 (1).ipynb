```python
def rotar(img, grados):
    h, w = img.shape[:2]
    M = cv2.getRotationMatrix2D((w//2, h//2), grados, 1.0)
    return cv2.warpAffine(img, M, (w, h))

def cambiar_brillo(img, factor):
    pil = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    pil = ImageEnhance.Brightness(pil).enhance(factor)
    return cv2.cvtColor(np.array(pil), cv2.COLOR_RGB2BGR)

def zoom(img, factor):
    h, w = img.shape[:2]
    nueva_h, nueva_w = int(h * factor), int(w * factor)
    resized = cv2.resize(img, (nueva_w, nueva_h))
    # Recortar o rellenar para mantener tamaño original
    if factor > 1:
        inicio_y = (nueva_h - h) // 2
        inicio_x = (nueva_w - w) // 2
        return resized[inicio_y:inicio_y+h, inicio_x:inicio_x+w]
    else:
        resultado = np.zeros_like(img)
        inicio_y = (h - nueva_h) // 2
        inicio_x = (w - nueva_w) // 2
        resultado[inicio_y:inicio_y+nueva_h, inicio_x:inicio_x+nueva_w] = resized
        return resultado

def desplazar(img, dx, dy):
    M = np.float32([[1, 0, dx], [0, 1, dy]])
    return cv2.warpAffine(img, M, (img.shape[1], img.shape[0]))

def agregar_ruido(img):
    ruido = np.random.normal(0, 15, img.shape).astype(np.int16)
    return np.clip(img.astype(np.int16) + ruido, 0, 255).astype(np.uint8)

def escala_grises_rgb(img):
    gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return cv2.cvtColor(gris, cv2.COLOR_GRAY2BGR)

def generar_variante(img):
    """Aplica transformaciones aleatorias combinadas a una imagen."""
    transformaciones = random.sample([
        lambda x: rotar(x, random.uniform(-15, 15)),
        lambda x: cambiar_brillo(x, random.uniform(0.7, 1.3)),
        lambda x: zoom(x, random.uniform(0.85, 1.15)),
        lambda x: desplazar(x, random.randint(-20, 20), random.randint(-20, 20)),
        lambda x: agregar_ruido(x),
        lambda x: escala_grises_rgb(x),
        lambda x: cv2.flip(x, 1),  # flip horizontal
    ], k=random.randint(2, 4))  # aplica entre 2 y 4 transformaciones por variante

    for t in transformaciones:
        img = t(img)
    return img

print(' Funciones, de transformación definidas')
print('   Transformaciones disponibles:')
print('  Rotacioon ±15°')
print('    Brillo ±30%')
print('    Zoom 85–115%')
print('    Desplazamiento')
print('    Ruido gaussiano')
print('    Escala de grises')
print('    Flip horizontal')
```

```python
VARIANTES_POR_IMAGEN = 50  # 4 fotos × 50 = 200 por alumno → 4,000 total

print(' Vamso Iniciando Data Augmentation...')
print(f'   {VARIANTES_POR_IMAGEN} variantes por imagen × 4 fotos × 20 alumnos = {VARIANTES_POR_IMAGEN*4*20} imágenes\n')

total_generadas = 0

for alumno in alumnos:
    carpeta = alumno['carpeta']
    carpeta_aug = f'{AUG}/{carpeta}'
    os.makedirs(carpeta_aug, exist_ok=True)

    fotos = ['frente_1.jpeg', 'frente_2.jpeg', 'perfil_izq.jpeg', 'perfil_der.jpeg']
    generadas_alumno = 0

    for foto in fotos:
        ruta_orig = f'{RAW}/{carpeta}/{foto}'
        img = cv2.imread(ruta_orig)

        if img is None:
            print(f'   ⚠️  No se pudo leer: {ruta_orig}')
            continue

        nombre_base = foto.replace('.jpeg', '')

        for i in range(VARIANTES_POR_IMAGEN):
            variante = generar_variante(img)
            nombre_out = f'{nombre_base}_aug_{i+1:03d}.jpeg'
            ruta_out = f'{carpeta_aug}/{nombre_out}'
            cv2.imwrite(ruta_out, variante)
            generadas_alumno += 1

    total_generadas += generadas_alumno
    print(f'    {carpeta:<25} {generadas_alumno} imágenes generadas')

print(f'\n Data Augmentation completada')
print(f'   Total imágenes generadas: {total_generadas}')
```

```python
print(' Resumen del dataset aumentado:\n')

total = 0
for alumno in alumnos:
    carpeta = alumno['carpeta']
    carpeta_aug = f'{AUG}/{carpeta}'
    n = len([f for f in os.listdir(carpeta_aug) if f.endswith('.jpeg')])
    total += n
    estado = 'bien' if n == 200 else f'⚠️  ({n}/200)'
    print(f'   {estado} {carpeta:<25} {n} imágenes')

print(f'\n   Total en dataset aumemtado: {total} imágenes')
print(f'   Imágenes originales (raw):  80 imágenes')
print(f'   Total combinado:            {total + 80} imágenes')
```
